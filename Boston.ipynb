```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from keras.models import Sequential
from keras.layers import Dense
```

```python
df = pd.read_csv("1_boston_housing.csv")
df.shape
```

```python
X = df.loc[:,['LSTAT','RM','PTRATIO']]
Y = df.loc[:,["MEDV"]]
X.shape, Y.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size = 0.25, random_state = 10)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train = scaler.transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = Sequential()
```

```python
model.add(Dense(128, input_shape=(3,), activation='relu', name= 'input'))
model.add(Dense(64, activation='relu', name='layer_1'))
model.add(Dense(1, activation='linear', name='output'))
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()
```

```python
model.fit(x_train, y_train, epochs= 100, validation_split= 0.05)
```

```python
output = model.evaluate(x_test,y_test)
```

```python
print(f"Mean Squared Error: {output[0]}"
     ,f"Mean Absolute Error: {output[1]}", sep="\n")
```

```python
y_pred = model.predict(x_test).flatten()
y_test = y_test.values.flatten()
```

```python
for i in range(5):
    print(f"Actual: {y_test[i]:.2f} | Predicted: {y_pred[i]:.2f}")
```
